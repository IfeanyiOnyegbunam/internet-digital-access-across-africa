```python
import pandas as pd
from pathlib import Path

# ---------------------------------------------------------
# 1. File paths
# ---------------------------------------------------------

input_file = Path(
    r"C:\ProgramData\MySQL\MySQL Server 8.0\Uploads\WDICSV.csv"
)

output_file = Path(
    r"C:\Users\pc\OneDrive\Documents\Data Analytics Projects\AnalystLab Africa Internship\Datasets\WDI_Technology_Africa_Prepared.csv"
)

# ---------------------------------------------------------
# 2. Selected indicators
# ---------------------------------------------------------

selected_indicators = [
    "IT.NET.USER.ZS",      # Individuals using the Internet (% of population)
    "IT.CEL.SETS.P2",      # Mobile cellular subscriptions (per 100 people)
    "IT.NET.BBND.P2",      # Fixed broadband subscriptions (per 100 people)
    "IT.NET.SECR.P6"       # Secure Internet servers (per 1 million people)
]

# ---------------------------------------------------------
# 3. Years
# ---------------------------------------------------------

years = [str(year) for year in range(2000, 2026)]

columns_to_keep = [
    "Country Name",
    "Country Code",
    "Indicator Name",
    "Indicator Code"
] + years

# ---------------------------------------------------------
# 4. Process the large WDI file in chunks
# ---------------------------------------------------------

chunk_size = 50000
processed_chunks = []

print("Starting WDI processing...")

for i, chunk in enumerate(
    pd.read_csv(
        input_file,
        usecols=columns_to_keep,
        chunksize=chunk_size,
        low_memory=False
    )
):
    
    # Keep only the four selected indicators
    chunk = chunk[
        chunk["Indicator Code"].isin(selected_indicators)
    ].copy()
    
    if chunk.empty:
        continue
    
    # Convert wide format to long format
    chunk = chunk.melt(
        id_vars=[
            "Country Name",
            "Country Code",
            "Indicator Name",
            "Indicator Code"
        ],
        value_vars=years,
        var_name="Year",
        value_name="Value"
    )
    
    processed_chunks.append(chunk)
    
    print(f"Processed chunk {i + 1}")

# ---------------------------------------------------------
# 5. Combine processed chunks
# ---------------------------------------------------------

df = pd.concat(processed_chunks, ignore_index=True)

# ---------------------------------------------------------
# 6. Clean values
# ---------------------------------------------------------

df["Year"] = pd.to_numeric(df["Year"], errors="coerce")

df["Value"] = pd.to_numeric(
    df["Value"],
    errors="coerce"
)

# Remove missing indicator values
df = df.dropna(subset=["Value"])

# ---------------------------------------------------------
# 7. Sort the dataset
# ---------------------------------------------------------

df = df.sort_values(
    ["Country Name", "Indicator Code", "Year"]
).reset_index(drop=True)

# ---------------------------------------------------------
# 8. Save the processed dataset
# ---------------------------------------------------------

df.to_csv(
    output_file,
    index=False
)

# ---------------------------------------------------------
# 9. Display results
# ---------------------------------------------------------

print("\nProcessing complete!")
print(f"Rows: {len(df):,}")
print(f"Columns: {len(df.columns)}")
print(f"Output file: {output_file}")

display(df.head(10))
```

```python
# ---------------------------------------------------------
# STEP 4: Identify the African countries
# ---------------------------------------------------------
# The World Bank places Sub-Saharan African countries under
# "Sub-Saharan Africa", while Egypt is classified under
# "Middle East & North Africa".
#
# We therefore use:
#   1. All countries in Sub-Saharan Africa
#   2. Egypt, which is geographically in Africa
#
# We do NOT include the other MENA countries because they
# are geographically outside Africa.
# ---------------------------------------------------------

# Get all Sub-Saharan African countries
ssa_countries = countries[
    countries["Region"] == "Sub-Saharan Africa"
].copy()

# Add Egypt separately because it is classified as MENA
# by the World Bank but is geographically an African country.
egypt = countries[
    countries["Country Code"] == "EGY"
].copy()

# Combine the two groups
africa_countries = pd.concat(
    [ssa_countries, egypt],
    ignore_index=True
)

# Keep only the fields we need for our project
africa_countries = africa_countries[
    [
        "Country Code",
        "Short Name",
        "Region",
        "Income Group"
    ]
].copy()

# Rename columns to make them easier to understand
africa_countries = africa_countries.rename(
    columns={
        "Short Name": "Country"
    }
)

# Remove any accidental duplicate country codes
africa_countries = africa_countries.drop_duplicates(
    subset="Country Code"
)

# Standardize Country names
africa_countries["Country"] = africa_countries["Country"].replace(
    {"Egypt, Arab Republic of": "Egypt",
     "Congo, Republic of": "Rep. Congo",
     "Congo, Democratic Republic of" : "Dem. Rep. Congo",
     "Somalia, Fed. Rep." : "Somalia",
     "Gambia, The" : "Gambia"}
)

# Sort alphabetically
africa_countries = africa_countries.sort_values(
    "Country"
).reset_index(drop=True)

# Show the result
print(f"Number of African countries identified: {len(africa_countries)}")

display(africa_countries)
```

```python
# ---------------------------------------------------------
# STEP 5: Load and filter the technology dataset
# ---------------------------------------------------------
# Our previous Python process converted the original WDI
# wide-format dataset into a long-format dataset.
#
# Now we:
#   1. Load that prepared dataset
#   2. Keep only our 49 African countries
#   3. Add country metadata such as income group
#   4. Validate the resulting dataset
#
# We join using Country Code because it is a reliable
# identifier and avoids problems caused by country names.
# ---------------------------------------------------------

# Path to the technology dataset we created earlier
technology_file = Path(
    r"C:\Users\pc\OneDrive\Documents\Data Analytics Projects\AnalystLab Africa Internship\Datasets\WDI_Technology_Africa_Prepared.csv"
)

# Load the prepared technology data
technology = pd.read_csv(technology_file)

print("Technology dataset loaded successfully.")
print(f"Rows: {len(technology):,}")
print(f"Columns: {len(technology.columns)}")

# ---------------------------------------------------------
# Check the columns before we continue
# ---------------------------------------------------------

print("\nColumns:")
print(technology.columns.tolist())

# ---------------------------------------------------------
# Filter the technology data using our African country list
# ---------------------------------------------------------
# Only country codes appearing in africa_countries
# will be retained.
# ---------------------------------------------------------

africa_technology = technology[
    technology["Country Code"].isin(
        africa_countries["Country Code"]
    )
].copy()

print(
    f"\nRows after filtering to African countries: "
    f"{len(africa_technology):,}"
)

# ---------------------------------------------------------
# Add country metadata
# ---------------------------------------------------------
# This gives us useful dimensions for later analysis:
#   Country
#   Region
#   Income Group
# ---------------------------------------------------------

africa_technology = africa_technology.merge(
    africa_countries,
    on="Country Code",
    how="left"
)

# ---------------------------------------------------------
# Check whether every technology record matched a country
# ---------------------------------------------------------

unmatched = africa_technology[
    africa_technology["Country"].isna()
]

print(
    f"\nUnmatched records after country join: "
    f"{len(unmatched):,}"
)

# ---------------------------------------------------------
# Display the first few records
# ---------------------------------------------------------

display(africa_technology.head(10))

# ---------------------------------------------------------
# Basic validation
# ---------------------------------------------------------

print("\nUnique African countries in dataset:")
print(africa_technology["Country"].nunique())

print("\nIndicators:")
print(
    africa_technology[
        ["Indicator Code", "Indicator Name"]
    ].drop_duplicates().to_string(index=False)
)

print("\nYear range:")
print(
    africa_technology["Year"].min(),
    "to",
    africa_technology["Year"].max()
)
```

```python
# ---------------------------------------------------------
# CHECK 2025 DATA AVAILABILITY
# ---------------------------------------------------------
# We want to know whether 2025 actually contains values
# for our four selected indicators in the African countries.
# ---------------------------------------------------------

print("2025 records:")
display(
    africa_technology[
        africa_technology["Year"] == 2025
    ]
)

print(
    "\nNumber of non-missing 2025 records:",
    africa_technology[
        africa_technology["Year"] == 2025
    ]["Value"].notna().sum()
)

# ---------------------------------------------------------
# CHECK THE LATEST AVAILABLE YEAR FOR EACH INDICATOR
# ---------------------------------------------------------
# Different WDI indicators may have different data coverage.
# This tells us exactly how far each indicator currently goes.
# ---------------------------------------------------------

latest_year_by_indicator = (
    africa_technology
    .groupby(
        ["Indicator Code", "Indicator Name"]
    )["Year"]
    .max()
    .reset_index()
    .sort_values("Indicator Code")
)

display(latest_year_by_indicator)
```

```python
# ---------------------------------------------------------
# STEP 6: DATA COVERAGE AND MISSINGNESS CHECK
# ---------------------------------------------------------
# Before analysing trends or ranking countries, we need to
# understand how complete our data is.
#
# This check helps us identify:
#   - How many observations each indicator has
#   - The available year range for each indicator
#   - How many countries have data for each indicator
#   - Which indicators have the most missing data
# ---------------------------------------------------------

# ---------------------------------------------------------
# 1. Overall coverage by indicator
# ---------------------------------------------------------
# This tells us how many records we have for each indicator
# and the first and last available year.
# ---------------------------------------------------------

indicator_coverage = (
    africa_technology
    .groupby(
        ["Indicator Code", "Indicator Name"]
    )
    .agg(
        Records=("Value", "count"),
        Countries=("Country Code", "nunique"),
        First_Year=("Year", "min"),
        Last_Year=("Year", "max")
    )
    .reset_index()
)

print("Coverage by indicator:")
display(indicator_coverage)


# ---------------------------------------------------------
# 2. Number of observations by country
# ---------------------------------------------------------
# This helps us identify countries with unusually limited
# data coverage.
# ---------------------------------------------------------

country_coverage = (
    africa_technology
    .groupby(
        ["Country Code", "Country"]
    )
    .agg(
        Records=("Value", "count"),
        Indicators=("Indicator Code", "nunique"),
        First_Year=("Year", "min"),
        Last_Year=("Year", "max")
    )
    .reset_index()
    .sort_values("Records")
)

print("\nCountry-level coverage:")
display(country_coverage)


# ---------------------------------------------------------
# 3. Check how many countries have each indicator
# ---------------------------------------------------------
# Ideally, we want to know whether all 49 countries appear
# for each indicator.
# ---------------------------------------------------------

country_indicator_coverage = (
    africa_technology
    .groupby("Indicator Code")["Country Code"]
    .nunique()
    .reset_index(name="Countries_With_Data")
)

print("\nCountries with data for each indicator:")
display(country_indicator_coverage)


# ---------------------------------------------------------
# 4. Check for duplicate records
# ---------------------------------------------------------
# A country + indicator + year combination should represent
# one observation.
#
# Duplicate records could distort our analysis, so we check
# for them before continuing.
# ---------------------------------------------------------

duplicates = (
    africa_technology
    .groupby(
        ["Country Code", "Indicator Code", "Year"]
    )
    .size()
    .reset_index(name="Count")
)

duplicates = duplicates[
    duplicates["Count"] > 1
]

print(
    f"\nDuplicate Country + Indicator + Year records: "
    f"{len(duplicates)}"
)

if len(duplicates) > 0:
    display(duplicates.head(20))
```

```python
# ---------------------------------------------------------
# STEP 7: IDENTIFY MISSING YEARS
# ---------------------------------------------------------
# We know that all 49 countries have all four indicators.
# However, some countries have fewer yearly observations.
#
# This step identifies:
#   1. How many observations each country has for each
#      indicator
#   2. Which years are missing
#   3. Whether missingness is concentrated in particular
#      indicators or countries
# ---------------------------------------------------------

# ---------------------------------------------------------
# Create a country-indicator coverage table
# ---------------------------------------------------------
# Each row represents one country and one indicator.
# "Years_Available" tells us how many years contain data.
# ---------------------------------------------------------

coverage_detail = (
    africa_technology
    .groupby(
        [
            "Country Code",
            "Country",
            "Indicator Code"
        ]
    )
    .agg(
        Years_Available=("Year", "nunique"),
        First_Year=("Year", "min"),
        Last_Year=("Year", "max")
    )
    .reset_index()
)

# Add the indicator name for easier reading
indicator_names = (
    africa_technology[
        ["Indicator Code", "Indicator Name"]
    ]
    .drop_duplicates()
)

coverage_detail = coverage_detail.merge(
    indicator_names,
    on="Indicator Code",
    how="left"
)

# Put the columns in a logical order
coverage_detail = coverage_detail[
    [
        "Country Code",
        "Country",
        "Indicator Code",
        "Indicator Name",
        "Years_Available",
        "First_Year",
        "Last_Year"
    ]
]

print("Country + Indicator coverage:")
display(coverage_detail)


# ---------------------------------------------------------
# Find the countries with the lowest data coverage
# ---------------------------------------------------------
# This helps us identify countries that may require
# additional caution during trend analysis.
# ---------------------------------------------------------

lowest_coverage = (
    coverage_detail
    .sort_values(
        ["Years_Available", "Country"]
    )
)

print("\nLowest country-indicator coverage:")
display(lowest_coverage.head(20))


# ---------------------------------------------------------
# Count observations by year and indicator
# ---------------------------------------------------------
# This shows how many countries have data in each year.
#
# If the number suddenly falls in a particular year,
# we know that year has incomplete country coverage.
# ---------------------------------------------------------

year_coverage = (
    africa_technology
    .groupby(
        ["Indicator Code", "Year"]
    )
    .agg(
        Countries_With_Data=("Country Code", "nunique")
    )
    .reset_index()
)

# Add indicator names
year_coverage = year_coverage.merge(
    indicator_names,
    on="Indicator Code",
    how="left"
)

year_coverage = year_coverage[
    [
        "Indicator Code",
        "Indicator Name",
        "Year",
        "Countries_With_Data"
    ]
]

print("\nCountry coverage by year:")
display(year_coverage)
```

```python
# ============================================================
# FINAL WDI TECHNOLOGY DATASET
# AFRICA-ONLY + COUNTRY INFORMATION
# ============================================================
# Goal:
# Create the final clean dataset that we will use in Power BI.
#
# The final dataset should contain:
# - Only African countries
# - 4 technology indicators
# - Available years and values
# - Country information
# - Region
# - Income Group
#
# Expected result:
# Approximately 3,975 rows
# ============================================================


import pandas as pd


# ------------------------------------------------------------
# 1. FILE PATHS
# ------------------------------------------------------------
# We are using:
#
# 1. The cleaned long-format technology dataset
# 2. The country lookup table containing Region and Income Group
#
# The original WDI data remains untouched.
# ------------------------------------------------------------

technology_path = r"C:\Users\pc\OneDrive\Documents\Data Analytics Projects\AnalystLab Africa Internship\Datasets\WDI_Technology_Africa_Prepared.csv"

country_lookup_path = r"C:\Users\pc\OneDrive\Documents\Data Analytics Projects\AnalystLab Africa Internship\Datasets\WDICountry.csv"


# ------------------------------------------------------------
# 2. LOAD THE TECHNOLOGY DATA
# ------------------------------------------------------------
# This file contains the WDI technology indicators in
# long format:
#
# Country | Indicator | Year | Value
#
# Long format is easier to analyse in Power BI.
# ------------------------------------------------------------

technology = pd.read_csv(technology_path)

print("Technology dataset loaded.")
print(f"Rows: {len(technology):,}")
print(f"Columns: {len(technology.columns)}")
print()


# ------------------------------------------------------------
# 3. LOAD THE COUNTRY LOOKUP
# ------------------------------------------------------------
# WDICountry contains information about countries including
# their region and income classification.
# ------------------------------------------------------------

countries = pd.read_csv(country_lookup_path)

print("Country lookup loaded.")
print(f"Rows: {len(countries):,}")
print(f"Columns: {len(countries.columns)}")
print()


# ------------------------------------------------------------
# 4. CHECK THE COUNTRY LOOKUP COLUMNS
# ------------------------------------------------------------
# We need to identify the exact columns containing:
# - Country Code
# - Short Name
# - Region
# - Income Group
#
# This check prevents us from guessing column names.
# ------------------------------------------------------------

print("Country lookup columns:")
print(countries.columns.tolist())
print()


# ------------------------------------------------------------
# 5. KEEP ONLY THE COUNTRY INFORMATION WE NEED
# ------------------------------------------------------------
# We only need these four columns from the country lookup.
# ------------------------------------------------------------

country_info = countries[
    ["Country Code", "Short Name", "Region", "Income Group"]
].copy()


# ------------------------------------------------------------
# 6. FILTER TO AFRICAN COUNTRIES
# ------------------------------------------------------------
# The World Bank region field identifies Sub-Saharan Africa,
# but North Africa is classified under Middle East &
# North Africa.
#
# Therefore, we cannot simply filter Region =
# "Sub-Saharan Africa".
#
# Instead, we use our verified list of 49 African country
# codes from the Africa country lookup we created earlier.
# ------------------------------------------------------------

africa_codes = [
    "AGO", "BEN", "BWA", "BFA", "BDI", "CPV", "CMR",
    "CAF", "TCD", "COM", "COD", "COG", "CIV", "EGY",
    "GNQ", "ERI", "SWZ", "ETH", "GAB", "GMB", "GHA",
    "GIN", "GNB", "KEN", "LSO", "LBR", "MDG", "MWI",
    "MLI", "MRT", "MUS", "MOZ", "NAM", "NER", "NGA",
    "RWA", "SEN", "SYC", "SLE", "SOM", "ZAF", "SSD",
    "SDN", "STP", "TZA", "TGO", "UGA", "ZMB", "ZWE"
]

print(f"African country codes defined: {len(africa_codes)}")
print()


# ------------------------------------------------------------
# 7. FILTER THE TECHNOLOGY DATA TO AFRICAN COUNTRIES
# ------------------------------------------------------------
# Only observations belonging to our 49 African countries
# should remain in the final analytical dataset.
# ------------------------------------------------------------

africa_technology = technology[
    technology["Country Code"].isin(africa_codes)
].copy()

print("After filtering to African countries:")
print(f"Rows: {len(africa_technology):,}")
print(f"Unique countries: {africa_technology['Country Code'].nunique()}")
print()


# ------------------------------------------------------------
# 8. JOIN COUNTRY INFORMATION
# ------------------------------------------------------------
# We now attach:
# - Country name
# - Region
# - Income Group
#
# Country Code is used as the matching key because it is a
# standardized identifier and is more reliable than country
# names.
# ------------------------------------------------------------

final_df = africa_technology.merge(
    country_info,
    on="Country Code",
    how="left"
)


# ------------------------------------------------------------
# 9. REMOVE THE DUPLICATE COUNTRY NAME COLUMN
# ------------------------------------------------------------
# The technology dataset already has "Country Name".
# The lookup has "Short Name".
#
# We don't need both, so we remove the lookup's duplicate
# country-name field.
# ------------------------------------------------------------

final_df = final_df.drop(columns=["Short Name"])


# ------------------------------------------------------------
# 10. CHECK THE JOIN
# ------------------------------------------------------------
# If Region or Income Group is missing, it means a country
# could not be matched to the lookup table.
#
# We expect zero unmatched records.
# ------------------------------------------------------------

missing_region = final_df["Region"].isna().sum()
missing_income = final_df["Income Group"].isna().sum()

print("Join validation:")
print(f"Missing Region: {missing_region}")
print(f"Missing Income Group: {missing_income}")
print()


# ------------------------------------------------------------
# 11. CHECK FOR DUPLICATES
# ------------------------------------------------------------
# A country should have only one observation for each
# indicator and year.
#
# For example:
# Angola + Mobile subscriptions + 2020
#
# should appear only once.
# ------------------------------------------------------------

duplicates = final_df.duplicated(
    subset=["Country Code", "Indicator Code", "Year"]
).sum()

print(f"Duplicate Country + Indicator + Year records: {duplicates}")
print()


# ------------------------------------------------------------
# 12. CHECK THE YEAR RANGE
# ------------------------------------------------------------
# We are checking the actual years that contain observations.
#
# Remember:
# Missing years are NOT automatically zeros.
# They simply mean the World Bank did not provide an
# observation for that country/indicator/year.
# ------------------------------------------------------------

print(f"Earliest year: {final_df['Year'].min()}")
print(f"Latest year: {final_df['Year'].max()}")
print()


# ------------------------------------------------------------
# 13. CHECK INDICATORS
# ------------------------------------------------------------
# Confirm that our four technology indicators are present.
# ------------------------------------------------------------

print("Indicators:")
print(
    final_df[
        ["Indicator Code", "Indicator Name"]
    ].drop_duplicates().to_string(index=False)
)
print()


# ------------------------------------------------------------
# 14. FINAL DATASET SUMMARY
# ------------------------------------------------------------

print("============================================================")
print("FINAL DATASET SUMMARY")
print("============================================================")

print(f"Rows:              {len(final_df):,}")
print(f"Columns:           {len(final_df.columns)}")
print(f"Countries:         {final_df['Country Code'].nunique()}")
print(f"Indicators:        {final_df['Indicator Code'].nunique()}")
print(f"Year range:        {final_df['Year'].min()}–{final_df['Year'].max()}")
print(f"Duplicates:        {duplicates}")
print(f"Missing Region:    {missing_region}")
print(f"Missing Income:    {missing_income}")

print("============================================================")
print()


# ------------------------------------------------------------
# 15. DISPLAY THE FINAL DATASET
# ------------------------------------------------------------
# Looking at the first few rows helps us visually confirm
# that everything looks correct before saving the file.
# ------------------------------------------------------------

display(final_df.head(10))


# ------------------------------------------------------------
# 16. SAVE THE FINAL DATASET
# ------------------------------------------------------------
# This is the file we will eventually import into Power BI.
#
# We are deliberately giving it a new name so we don't
# accidentally overwrite the previous intermediate file.
# ------------------------------------------------------------

output_path = r"C:\Users\pc\OneDrive\Documents\Data Analytics Projects\AnalystLab Africa Internship\Datasets\WDI_Technology_Africa_Final.csv"

final_df.to_csv(
    output_path,
    index=False
)

print()
print("============================================================")
print("FINAL DATASET SAVED SUCCESSFULLY")
print("============================================================")
print(f"Output file: {output_path}")
```

```python
# ============================================================
# FINAL WDI TECHNOLOGY DATASET
# AFRICA-ONLY + COUNTRY INFORMATION
# ============================================================
# Goal:
# Create the final clean dataset that we will use in Power BI.
#
# The final dataset should contain:
# - Only African countries
# - 4 technology indicators
# - Available years and values
# - Country information
# - Region
# - Income Group
#
# Expected result:
# Approximately 3,975 rows
# ============================================================


import pandas as pd


# ------------------------------------------------------------
# 1. FILE PATHS
# ------------------------------------------------------------
# We are using:
#
# 1. The cleaned long-format technology dataset
# 2. The country lookup table containing Region and Income Group
#
# The original WDI data remains untouched.
# ------------------------------------------------------------

technology_path = r"C:\Users\pc\OneDrive\Documents\Data Analytics Projects\AnalystLab Africa Internship\Datasets\WDI_Technology_Africa_Prepared.csv"

country_lookup_path = r"C:\Users\pc\OneDrive\Documents\Data Analytics Projects\AnalystLab Africa Internship\Datasets\WDICountry.csv"


# ------------------------------------------------------------
# 2. LOAD THE TECHNOLOGY DATA
# ------------------------------------------------------------
# This file contains the WDI technology indicators in
# long format:
#
# Country | Indicator | Year | Value
#
# Long format is easier to analyse in Power BI.
# ------------------------------------------------------------

technology = pd.read_csv(technology_path)

print("Technology dataset loaded.")
print(f"Rows: {len(technology):,}")
print(f"Columns: {len(technology.columns)}")
print()


# ------------------------------------------------------------
# 3. LOAD THE COUNTRY LOOKUP
# ------------------------------------------------------------
# WDICountry contains information about countries including
# their region and income classification.
# ------------------------------------------------------------

countries = pd.read_csv(country_lookup_path)

print("Country lookup loaded.")
print(f"Rows: {len(countries):,}")
print(f"Columns: {len(countries.columns)}")
print()


# ------------------------------------------------------------
# 4. CHECK THE COUNTRY LOOKUP COLUMNS
# ------------------------------------------------------------
# We need to identify the exact columns containing:
# - Country Code
# - Short Name
# - Region
# - Income Group
#
# This check prevents us from guessing column names.
# ------------------------------------------------------------

print("Country lookup columns:")
print(countries.columns.tolist())
print()


# ------------------------------------------------------------
# 5. KEEP ONLY THE COUNTRY INFORMATION WE NEED
# ------------------------------------------------------------
# We only need these four columns from the country lookup.
# ------------------------------------------------------------

country_info = countries[
    ["Country Code", "Short Name", "Region", "Income Group"]
].copy()


# ------------------------------------------------------------
# 6. FILTER TO AFRICAN COUNTRIES
# ------------------------------------------------------------
# The World Bank region field identifies Sub-Saharan Africa,
# but North Africa is classified under Middle East &
# North Africa.
#
# Therefore, we cannot simply filter Region =
# "Sub-Saharan Africa".
#
# Instead, we use our verified list of 49 African country
# codes from the Africa country lookup we created earlier.
# ------------------------------------------------------------

africa_codes = [
    "AGO", "BEN", "BWA", "BFA", "BDI", "CPV", "CMR",
    "CAF", "TCD", "COM", "COD", "COG", "CIV", "EGY",
    "GNQ", "ERI", "SWZ", "ETH", "GAB", "GMB", "GHA",
    "GIN", "GNB", "KEN", "LSO", "LBR", "MDG", "MWI",
    "MLI", "MRT", "MUS", "MOZ", "NAM", "NER", "NGA",
    "RWA", "SEN", "SYC", "SLE", "SOM", "ZAF", "SSD",
    "SDN", "STP", "TZA", "TGO", "UGA", "ZMB", "ZWE"
]

print(f"African country codes defined: {len(africa_codes)}")
print()


# ------------------------------------------------------------
# 7. FILTER THE TECHNOLOGY DATA TO AFRICAN COUNTRIES
# ------------------------------------------------------------
# Only observations belonging to our 49 African countries
# should remain in the final analytical dataset.
# ------------------------------------------------------------

africa_technology = technology[
    technology["Country Code"].isin(africa_codes)
].copy()

print("After filtering to African countries:")
print(f"Rows: {len(africa_technology):,}")
print(f"Unique countries: {africa_technology['Country Code'].nunique()}")
print()


# ------------------------------------------------------------
# 8. JOIN COUNTRY INFORMATION
# ------------------------------------------------------------
# We now attach:
# - Country name
# - Region
# - Income Group
#
# Country Code is used as the matching key because it is a
# standardized identifier and is more reliable than country
# names.
# ------------------------------------------------------------

final_df = africa_technology.merge(
    country_info,
    on="Country Code",
    how="left"
)


# ------------------------------------------------------------
# 9. REMOVE THE DUPLICATE COUNTRY NAME COLUMN
# ------------------------------------------------------------
# The technology dataset already has "Country Name".
# The lookup has "Short Name".
#
# We don't need both, so we remove the lookup's duplicate
# country-name field.
# ------------------------------------------------------------

final_df = final_df.drop(columns=["Short Name"])


# ------------------------------------------------------------
# 10. CHECK THE JOIN
# ------------------------------------------------------------
# If Region or Income Group is missing, it means a country
# could not be matched to the lookup table.
#
# We expect zero unmatched records.
# ------------------------------------------------------------

missing_region = final_df["Region"].isna().sum()
missing_income = final_df["Income Group"].isna().sum()

print("Join validation:")
print(f"Missing Region: {missing_region}")
print(f"Missing Income Group: {missing_income}")
print()


# ------------------------------------------------------------
# 11. CHECK FOR DUPLICATES
# ------------------------------------------------------------
# A country should have only one observation for each
# indicator and year.
#
# For example:
# Angola + Mobile subscriptions + 2020
#
# should appear only once.
# ------------------------------------------------------------

duplicates = final_df.duplicated(
    subset=["Country Code", "Indicator Code", "Year"]
).sum()

print(f"Duplicate Country + Indicator + Year records: {duplicates}")
print()


# ------------------------------------------------------------
# 14. FINAL STANDARDIZATION
# ------------------------------------------------------------
final_df["Country Name"] = final_df["Country Name"].replace(
    {"Egypt, Arab Rep.": "Egypt",
     "Congo, Rep.": "Rep. Congo",
     "Congo, Dem. Rep." : "Dem. Rep. Congo",
     "Somalia, Fed. Rep." : "Somalia",
     "Gambia, The" : "Gambia"}
)

# ------------------------------------------------------------
# 12. CHECK THE YEAR RANGE
# ------------------------------------------------------------
# We are checking the actual years that contain observations.
#
# Remember:
# Missing years are NOT automatically zeros.
# They simply mean the World Bank did not provide an
# observation for that country/indicator/year.
# ------------------------------------------------------------

print(f"Earliest year: {final_df['Year'].min()}")
print(f"Latest year: {final_df['Year'].max()}")
print()


# ------------------------------------------------------------
# 13. CHECK INDICATORS
# ------------------------------------------------------------
# Confirm that our four technology indicators are present.
# ------------------------------------------------------------

print("Indicators:")
print(
    final_df[
        ["Indicator Code", "Indicator Name"]
    ].drop_duplicates().to_string(index=False)
)
print()


# ------------------------------------------------------------
# 14. FINAL DATASET SUMMARY
# ------------------------------------------------------------

print("============================================================")
print("FINAL DATASET SUMMARY")
print("============================================================")

print(f"Rows:              {len(final_df):,}")
print(f"Columns:           {len(final_df.columns)}")
print(f"Countries:         {final_df['Country Code'].nunique()}")
print(f"Indicators:        {final_df['Indicator Code'].nunique()}")
print(f"Year range:        {final_df['Year'].min()}–{final_df['Year'].max()}")
print(f"Duplicates:        {duplicates}")
print(f"Missing Region:    {missing_region}")
print(f"Missing Income:    {missing_income}")

print("============================================================")
print()


# ------------------------------------------------------------
# 15. DISPLAY THE FINAL DATASET
# ------------------------------------------------------------
# Looking at the first few rows helps us visually confirm
# that everything looks correct before saving the file.
# ------------------------------------------------------------

display(final_df.head(10))


# ------------------------------------------------------------
# 16. SAVE THE FINAL DATASET
# ------------------------------------------------------------
# This is the file we will eventually import into Power BI.
#
# We are deliberately giving it a new name so we don't
# accidentally overwrite the previous intermediate file.
# ------------------------------------------------------------

output_path = r"C:\Users\pc\OneDrive\Documents\Data Analytics Projects\AnalystLab Africa Internship\Datasets\WDI_Technology_Africa_Final.csv"

final_df.to_csv(
    output_path,
    index=False
)

print()
print("============================================================")
print("FINAL DATASET SAVED SUCCESSFULLY")
print("============================================================")
print(f"Output file: {output_path}")
```
